# Shape-trace through scaled dot-product attention

```python
import torch

print('cuda available:', torch.cuda.is_available())
```

```python
# Starter: fill in each shape in the comments.
batch, seq, d_head = 2, 4, 8
q = torch.randn(batch, seq, d_head, device='cuda' if torch.cuda.is_available() else 'cpu')
k = torch.randn(batch, seq, d_head, device=q.device)
v = torch.randn(batch, seq, d_head, device=q.device)

scores = q @ k.transpose(-1, -2)
weights = torch.softmax(scores / (d_head ** 0.5), dim=-1)
out = weights @ v

print('scores shape:', scores.shape)
print('weights shape:', weights.shape)
print('out shape:', out.shape)
```

```python
# Reveal solution
# q @ k^T -> (2, 4, 4)
# softmax -> (2, 4, 4)
# weights @ v -> (2, 4, 8)
```
